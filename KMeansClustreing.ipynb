```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import seaborn as sns
```

```python
data=pd.read_csv('Mall_Customers.csv')
data
```

```python
data=data.rename(columns={'Annual Income (k$)':'annInc','Spending Score (1-100)':'spendingSc'})
```

```python
plt.scatter(data['annInc'],data['spendingSc'])
```

```python
x=data.iloc[:,3:]
```

```python
km=KMeans(n_clusters=5)
y_means=km.fit_predict(x)

sns.scatterplot(data=data,x=data['annInc'],y=data['spendingSc'],hue=y_means)
plt.scatter(km.cluster_centers_[:,0],km.cluster_centers_[:,1],marker='*')

plt.show()
```
